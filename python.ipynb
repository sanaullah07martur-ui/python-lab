```python
# Q9. Factorial using Function

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

num = int(input("Enter number: "))
print("Factorial:", factorial(num))
```

```python
# Q10. Prime Number Checker

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not Prime")
```

```python
# Q15. Word Frequency Counter

sentence = input("Enter sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
# Q16. Library Book System

books = []

def add_book(book):
    books.append(book)

def remove_book(book):
    if book in books:
        books.remove(book)

def search_book(book):
    if book in books:
        print("Book Found")
    else:
        print("Book Not Found")

def display_books():
    print("Books:", books)

while True:
    print("\n1.Add Book")
    print("2.Remove Book")
    print("3.Search Book")
    print("4.Display Books")
    print("5.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        b = input("Enter book name: ")
        add_book(b)

    elif choice == 2:
        b = input("Enter book name: ")
        remove_book(b)

    elif choice == 3:
        b = input("Enter book name: ")
        search_book(b)

    elif choice == 4:
        display_books()

    elif choice == 5:
        break
```

```python
# Q17. Student Result Analyzer

marks = []

for i in range(5):
    m = int(input("Enter marks: "))
    marks.append(m)

def average(marks):
    return sum(marks) / len(marks)

def highest(marks):
    return max(marks)

def lowest(marks):
    return min(marks)

def passed_students(marks):
    count = 0
    for m in marks:
        if m > 40:
            count += 1
    return count

print("Average:", average(marks))
print("Highest:", highest(marks))
print("Lowest:", lowest(marks))
print("Passed Students:", passed_students(marks))
```

```python
# Q18. Password Strength Checker

def check_password(password):

    if len(password) < 8:
        return "Weak Password"

    has_digit = False
    has_upper = False
    has_lower = False

    for ch in password:
        if ch.isdigit():
            has_digit = True
        elif ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True

    if has_digit and has_upper and has_lower:
        return "Strong Password"
    else:
        return "Weak Password"

pwd = input("Enter password: ")
print(check_password(pwd))
```

```python
# Q19. Contact Book Application

contacts = {}

while True:
    print("\n1.Add Contact")
    print("2.Search Contact")
    print("3.Delete Contact")
    print("4.Display Contacts")
    print("5.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        name = input("Enter name: ")
        number = input("Enter number: ")
        contacts[name] = number

    elif choice == 2:
        name = input("Enter name: ")
        print(contacts.get(name, "Contact Not Found"))

    elif choice == 3:
        name = input("Enter name: ")
        if name in contacts:
            del contacts[name]

    elif choice == 4:
        print(contacts)

    elif choice == 5:
        break
```

```python
# Q21. HackerRank Style Problem

numbers = [10, 20, 30, 40, 50]

# Find maximum
maximum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num

# Find minimum
minimum = numbers[0]

for num in numbers:
    if num < minimum:
        minimum = num

# Reverse list
reversed_list = []

for i in range(len(numbers)-1, -1, -1):
    reversed_list.append(numbers[i])

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Reversed List:", reversed_list)
```
